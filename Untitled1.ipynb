```python
import pandas as pd

# ==========================================
# BƯỚC 1: THU THẬP & TÍCH HỢP DỮ LIỆU (PRODUCTION-READY)
# ==========================================

config = {
    "data_path" : {
        'train' : "train - Walmart Sales Forecast.csv",
        "test" : "test - Walmart Sales Forecast.csv",
        "stores" : "stores - Walmart Sales Forecast.csv",
        "features" : "features - Walmart Sales Forecast.csv",
    }
}

# 1. Parse trực tiếp Date ngay khi đọc CSV để tránh lệch kiểu object vs datetime64
df_train = pd.read_csv(config['data_path']['train'], parse_dates=['Date'])
df_test = pd.read_csv(config['data_path']['test'], parse_dates=['Date'])
df_features = pd.read_csv(config['data_path']['features'], parse_dates=['Date'])
df_stores = pd.read_csv(config['data_path']['stores'])

# 2. Xử lý ffill theo từng Store trên bảng features (tránh rò rỉ dữ liệu chéo)
df_features = df_features.sort_values(['Store', 'Date'])
economic_features = ['CPI', 'Unemployment']
df_features[economic_features] = df_features.groupby('Store')[economic_features].ffill()

# Loại bỏ cột IsHoliday thừa trong features để tránh xung đột
if 'IsHoliday' in df_features.columns:
    df_features.drop(columns=['IsHoliday'], inplace=True)

# 3. Hàm Merge an toàn với cơ chế Validate và Assert bảo toàn số dòng
def merge_data(df_base, df_features, df_stores):
    out = (df_base
           .merge(df_features, on=['Store', 'Date'], how='left', validate='m:1')
           .merge(df_stores, on='Store', how='left', validate='m:1'))
    
    assert len(out) == len(df_base), "Lỗi: Thao tác Merge làm thay đổi số lượng dòng dữ liệu gốc!"
    return out

df_train_full = merge_data(df_train, df_features, df_stores)
df_test_full = merge_data(df_test, df_features, df_stores)

print(df_train.shape, '->', df_train_full.shape)
print(df_test.shape, '->', df_test_full.shape)
```

```python
# ==============================================================================
# BƯỚC 1.2: TIỀN XỬ LÝ & KỸ NGHỆ ĐẶC TRƯNG DỰA TRÊN KẾT QUẢ KHÁM PHÁ
# ==============================================================================

# 1. Chuẩn hóa kiểu dữ liệu số cho biến logic IsHoliday
for frame in [df_train_full, df_test_full]:
    frame['IsHoliday'] = frame['IsHoliday'].astype(int)

# 2. Xử lý khuyết thiếu và tạo cờ nhận diện Covariate Shift cho MarkDown
markdown_cols = ['MarkDown1', 'MarkDown2', 'MarkDown3', 'MarkDown4', 'MarkDown5']
for frame in [df_train_full, df_test_full]:
    frame['has_markdown'] = frame[markdown_cols].notna().any(axis=1).astype(int)
    frame[markdown_cols] = frame[markdown_cols].fillna(0)

# 3. Kỹ nghệ đặc trưng Không gian - Thời gian
for frame in [df_train_full, df_test_full]:
    frame['Week'] = frame['Date'].dt.isocalendar().week.astype(int)
    frame['Month'] = frame['Date'].dt.month

# Mã hóa One-Hot cho biến phân loại Type (A, B, C)
df_train_full = pd.get_dummies(df_train_full, columns=['Type'], drop_first=True)
df_test_full = pd.get_dummies(df_test_full, columns=['Type'], drop_first=True)

# 4. Sắp xếp chuỗi thời gian và tạo đặc trưng trễ Lag_4 (Phục vụ dự báo 4 tuần tới)
df_train_full = df_train_full.sort_values(['Store', 'Dept', 'Date']).reset_index(drop=True)
df_test_full = df_test_full.sort_values(['Store', 'Dept', 'Date']).reset_index(drop=True)

df_train_full['Lag_4'] = df_train_full.groupby(['Store', 'Dept'])['Weekly_Sales'].shift(4)
df_train_full['Rolling_Mean_4'] = df_train_full.groupby(['Store', 'Dept'])['Weekly_Sales'].transform(
    lambda x: x.shift(4).rolling(window=4, min_periods=1).mean()
)

# Loại bỏ các dòng đầu bị thiếu do quá trình trượt cửa sổ (Lag_4 sinh ra 4 dòng NaN đầu mỗi nhóm)
df_train_full.dropna(subset=['Lag_4'], inplace=True)

# 5. Phân chia tập Validation 4 tuần cuối theo chiến lược Time-Series Split
max_date = df_train_full['Date'].max()
val_start_date = max_date - pd.Timedelta(weeks=3)

train_split = df_train_full[df_train_full['Date'] < val_start_date].copy()
val_split = df_train_full[df_train_full['Date'] >= val_start_date].copy()

# 6. Xử lý doanh số âm (Returns): Chỉ quy đổi về 0 trên tập Train
train_split.loc[train_split['Weekly_Sales'] < 0, 'Weekly_Sales'] = 0

print("--- KIỂM TRA TÍNH TOÀN VẸN SAU TIỀN XỬ LÝ ---")
print(f"Kích thước tập Train Split:      {train_split.shape}")
print(f"Kích thước tập Validation Split: {val_split.shape}")
assert train_split.isna().sum().sum() == 0, "Cảnh báo: Còn giá trị NaN trên Train Split!"
assert val_split.isna().sum().sum() == 0, "Cảnh báo: Còn giá trị NaN trên Validation Split!"
```

```python
# ==============================================================================
# BƯỚC 1.2: BÁO CÁO NGHIỆM THU CHẤT LƯỢNG DỮ LIỆU SAU TIỀN XỬ LÝ
# ==============================================================================

print("=" * 70)
print("             KẾT QUẢ TIỀN XỬ LÝ DỮ LIỆU")
print("=" * 70)

# 1. Kiểm toán kích thước và giá trị thiếu
print(f"1. KÍCH THƯỚC DỮ LIỆU:")
print(f"   - Tập Huấn luyện (Train Split): {train_split.shape[0]:,} dòng | {train_split.shape[1]} cột")
print(f"   - Tập Kiểm chứng (Val Split):   {val_split.shape[0]:,} dòng  | {val_split.shape[1]} cột")
print(f"   - Tổng số giá trị khuyết thiếu (Train): {train_split.isna().sum().sum()} (Đạt chuẩn 0 NaN)")
print(f"   - Tổng số giá trị khuyết thiếu (Val):   {val_split.isna().sum().sum()} (Đạt chuẩn 0 NaN)")

# 2. Kiểm toán trục thời gian (Time-Series Audit)
val_weeks = val_split['Date'].dt.strftime('%Y-%m-%d').unique().tolist()
print(f"\n2. KIỂM TOÁN CHUỖI THỜI GIAN:")
print(f"   - Train Split diễn ra từ: {train_split['Date'].min().strftime('%Y-%m-%d')} đến {train_split['Date'].max().strftime('%Y-%m-%d')} ({train_split['Date'].nunique()} tuần)")
print(f"   - Val Split diễn ra từ:   {val_split['Date'].min().strftime('%Y-%m-%d')} đến {val_split['Date'].max().strftime('%Y-%m-%d')} (Đúng {len(val_weeks)} tuần)")
print(f"   - Danh sách 4 mốc tuần kiểm chứng: {val_weeks}")

# 3. Kiểm toán xử lý doanh số âm (Returns)
print(f"\n3. KIỂM TOÁN BIẾN MỤC TIÊU (WEEKLY_SALES):")
print(f"   - Giá trị nhỏ nhất trên Train Split: {train_split['Weekly_Sales'].min():.2f} USD (Đã quy đổi >= 0)")
print(f"   - Giá trị nhỏ nhất trên Val Split:   {val_split['Weekly_Sales'].min():.2f} USD (Bảo lưu tính khách quan)")

# 4. Xác nhận hoàn tất
assert train_split.isna().sum().sum() == 0 and val_split.isna().sum().sum() == 0, "LỖI: Vẫn còn dữ liệu khuyết thiếu!"
assert len(val_weeks) == 4, "LỖI: Tập Validation không khớp 4 tuần!"
print("\n>>> KẾT LUẬN: Dữ liệu đạt chuẩn Production-Ready, sẵn sàng chuyển sang Bước 1.3 Lựa chọn mô hình.")
print("=" * 70)
```

```python
# ==============================================================================
# BƯỚC 1.2: CODE XÁC THỰC MINH CHỨNG SAU TIỀN XỬ LÝ (VERIFICATION SUITE)
# ==============================================================================

# --- MINH CHỨNG 1: XỬ LÝ DOANH SỐ ÂM (RETURNS) ---
print("=" * 75)
print("KẾT QUẢ XỬ LÝ DOANH SỐ ÂM (RETURNS) TRÊN TRAIN VS VALIDATION")
print("=" * 75)
proof_1 = pd.DataFrame({
    'Phân vùng': ['Tập Train (train_split)', 'Tập Validation (val_split)'],
    'Min Weekly_Sales': [f"{train_split['Weekly_Sales'].min():.2f} USD", f"{val_split['Weekly_Sales'].min():.2f} USD"],
    'Số lượng bản ghi < 0': [(train_split['Weekly_Sales'] < 0).sum(), (val_split['Weekly_Sales'] < 0).sum()],
    'Tỷ lệ bản ghi < 0': [f"{(train_split['Weekly_Sales'] < 0).mean() * 100:.3f}%", f"{(val_split['Weekly_Sales'] < 0).mean() * 100:.3f}%"],
    'Trạng thái xử lý': ['Đã quy đổi về 0.0 (Triệt tiêu nhiễu)', 'Bảo lưu nguyên trạng (Đo lường khách quan)']
})
display(proof_1)


# --- MINH CHỨNG 2: XỬ LÝ COVARIATE SHIFT (MARKDOWN & CỜ HAS_MARKDOWN) ---
print("\n" + "=" * 75)
print("KẾT QUẢ XỬ LÝ KHUYẾT THIẾU MARKDOWN & TẠO CỜ CHỈ BÁO HAS_MARKDOWN")
print("=" * 75)
md_cols = ['MarkDown1', 'MarkDown2', 'MarkDown3', 'MarkDown4', 'MarkDown5', 'has_markdown']
proof_2 = pd.DataFrame({
    'Thuộc tính': md_cols,
    'Số lượng NaN (Train)': [train_split[c].isna().sum() for c in md_cols],
    'Số lượng NaN (Val)': [val_split[c].isna().sum() for c in md_cols],
    'Giá trị Min': [round(train_split[c].min(), 2) for c in md_cols],
    'Giá trị Max': [round(train_split[c].max(), 2) for c in md_cols],
    'Kiểm toán sạch (0 NaN)': ['ĐẠT' if train_split[c].isna().sum() == 0 else 'CHƯA ĐẠT' for c in md_cols]
})
display(proof_2)


# --- MINH CHỨNG 3: MINH CHỨNG TOÁN HỌC LOGIC TỰ HỒI QUY LAG_4 ---
print("\n" + "=" * 75)
print("KẾT QUẢ TÍNH TOÁN ĐẶC TRƯNG LAG_4 VÀ ROLLING_MEAN_4 THEO CHUỖI THỜI GIAN")
print("=" * 75)
# Trích xuất 8 tuần đầu tiên của Store 1, Dept 1 để chứng minh logic shift(4)
sample_lag = df_train_full[(df_train_full['Store'] == 1) & (df_train_full['Dept'] == 1)][
    ['Date', 'Weekly_Sales', 'Lag_4', 'Rolling_Mean_4']
].head(8).copy()
sample_lag['Date'] = sample_lag['Date'].dt.strftime('%Y-%m-%d')
display(sample_lag)


# --- MINH CHỨNG 4: CHUẨN HÓA DỮ LIỆU & PHÂN CHIA TIME-SERIES SPLIT ĐÚNG 4 TUẦN ---
print("\n" + "=" * 75)
print("KẾT QUẢ CHUẨN HÓA KIỂU DỮ LIỆU VÀ PHÂN CHIA VALIDATION 4 TUẦN")
print("=" * 75)
encoded_cols = ['IsHoliday', 'Type_B', 'Type_C', 'Week', 'Month']
proof_4_types = pd.DataFrame({
    'Cột đặc trưng': encoded_cols,
    'Kiểu dữ liệu mới': [str(train_split[c].dtype) for c in encoded_cols],
    'Miền giá trị (Min - Max)': [f"{train_split[c].min()} đến {train_split[c].max()}" for c in encoded_cols],
    'Trạng thái': ['Chuẩn hóa số nguyên int' for _ in encoded_cols]
})
display(proof_4_types)

val_weeks = val_split['Date'].dt.strftime('%Y-%m-%d').unique().tolist()
print(f"- Số lượng tuần trong tập Validation Split: {len(val_weeks)} tuần")
print(f"- Danh sách cụ thể 4 mốc tuần kiểm chứng:   {val_weeks}")
print(f"- Kích thước tập Train: {train_split.shape[0]:,} dòng | Tập Validation: {val_split.shape[0]:,} dòng")
```

```python
import numpy as np 
from sklearn.linear_model import LinearRegression
feature_cols = [col for col in train_split.columns if col not in ['Date', 'Weekly_Sales']]
X_train = train_split[feature_cols]
y_train = train_split['Weekly_Sales']
X_val = val_split[feature_cols]
y_val = val_split['Weekly_Sales']

def calculate_wmae(y_true, y_pred, is_holiday):
    weights = np.where(is_holiday == 1, 5, 1)
    return np.sum(weights * np.abs(y_true - y_pred)) / np.sum(weights)

# Tính WMAE cho 3 Baseline
wmae_mean = calculate_wmae(y_val, np.full_like(y_val, y_train.mean()), X_val['IsHoliday'])
wmae_naive = calculate_wmae(y_val, X_val['Lag_4'], X_val['IsHoliday'])

lr = LinearRegression()
lr.fit(X_train, y_train)
wmae_lr = calculate_wmae(y_val, lr.predict(X_val), X_val['IsHoliday'])

print("\n" + "=" * 65)
print("TÍNH TOÁN SỐ LIỆU BƯỚC 1.3 (ĐẶC TRƯNG & BASELINE WMAE):")
print(f"- Số lượng đặc trưng đầu vào (X): {len(feature_cols)} biến")
print(f"- WMAE Mean Baseline:              {wmae_mean:,.2f} USD")
print(f"- WMAE Linear Regression:          {wmae_lr:,.2f} USD")
print(f"- WMAE Naïve Baseline (Lag_4):     {wmae_naive:,.2f} USD")
print("=" * 65)
```

```python
# ==============================================================================
# BƯỚC 1.4: HUẤN LUYỆN VÀ TINH CHỈNH MÔ HÌNH (TRAINING)
# ==============================================================================
import time
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor

print("BƯỚC 1.4: BẮT ĐẦU HUẤN LUYỆN CÁC MÔ HÌNH...")

# 1. Linear Regression
t0 = time.time()
lr_model = LinearRegression().fit(X_train, y_train)
lr_time = time.time() - t0
print(f"- Đã huấn luyện Linear Regression: {lr_time:.2f}s")

# 2. Random Forest
t0 = time.time()
rf_model = RandomForestRegressor(n_estimators=50, max_depth=12, random_state=42, n_jobs=-1).fit(X_train, y_train)
rf_time = time.time() - t0
print(f"- Đã huấn luyện Random Forest: {rf_time:.2f}s")

# 3. Gradient Boosting (Mặc định)
t0 = time.time()
gb_model = HistGradientBoostingRegressor(max_iter=150, max_depth=8, learning_rate=0.1, random_state=42).fit(X_train, y_train)
gb_time = time.time() - t0
print(f"- Đã huấn luyện Gradient Boosting (Mặc định): {gb_time:.2f}s")

# 4. Gradient Boosting (Tinh chỉnh siêu tham số)
t0 = time.time()
tuned_gb_model = HistGradientBoostingRegressor(
    max_iter=300, max_depth=10, learning_rate=0.08, l2_regularization=1.0, random_state=42
).fit(X_train, y_train)
tuned_time = time.time() - t0
print(f"- Đã huấn luyện Gradient Boosting (Tinh chỉnh): {tuned_time:.2f}s")
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Hàm tính WMAE (Phạt gấp 5 lần vào tuần lễ hội)
def calculate_wmae(y_true, y_pred, is_holiday):
    weights = np.where(is_holiday == 1, 5, 1)
    return np.sum(weights * np.abs(y_true - y_pred)) / np.sum(weights)

def evaluate_model(model_name, model, train_time):
    preds = model.predict(X_val)
    wmae = calculate_wmae(y_val, preds, X_val['IsHoliday'])
    mae = mean_absolute_error(y_val, preds)
    rmse = np.sqrt(mean_squared_error(y_val, preds))
    r2 = r2_score(y_val, preds)
    
    return {
        'Mô hình': model_name,
        'WMAE (USD)': f"{wmae:,.2f}",
        'MAE (USD)': f"{mae:,.2f}",
        'RMSE (USD)': f"{rmse:,.2f}",
        'R2 Score': f"{r2:.4f}",
        'Thời gian': f"{train_time:.2f}s"
    }

# Đo lường và đưa vào bảng
evaluation_results = [
    evaluate_model('Linear Regression', lr_model, lr_time),
    evaluate_model('Random Forest', rf_model, rf_time),
    evaluate_model('Gradient Boosting (Mặc định)', gb_model, gb_time),
    evaluate_model('Gradient Boosting (Tinh chỉnh)', tuned_gb_model, tuned_time)
]

print("\nBƯỚC 1.5: BẢNG TỔNG KẾT CHỈ SỐ ĐÁNH GIÁ TRÊN TẬP VALIDATION (4 TUẦN)")
df_evaluation = pd.DataFrame(evaluation_results)
display(df_evaluation)
```

```python
# ==============================================================================
# BƯỚC 3.6: GIẢI MÃ HỘP ĐEN MÔ HÌNH BẰNG PERMUTATION IMPORTANCE
# ==============================================================================
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.inspection import permutation_importance

print("Đang tính toán mức độ ảnh hưởng của các đặc trưng (Permutation Importance)...")

# Tính toán tầm quan trọng (Sử dụng tuned_gb_model từ Bước 5)
result = permutation_importance(
    tuned_gb_model, X_val, y_val, 
    scoring='neg_mean_absolute_error', # Đo lường lượng sai số MAE tăng lên nếu biến bị xáo trộn
    n_repeats=5, 
    random_state=42, 
    n_jobs=-1
)

# Tạo bảng dữ liệu
feature_importances = pd.DataFrame({
    'Đặc trưng': X_val.columns,
    'Tầm quan trọng': result.importances_mean
})
# Sắp xếp giảm dần
feature_importances = feature_importances.sort_values(by='Tầm quan trọng', ascending=False)

# Vẽ biểu đồ trực quan
sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(12, 8))

sns.barplot(
    data=feature_importances, 
    x='Tầm quan trọng', 
    y='Đặc trưng', 
    palette='viridis', 
    ax=ax
)

ax.set_title('Giải mã mô hình - Tầm quan trọng của các đặc trưng (Permutation Importance)', fontsize=14, fontweight='bold')
ax.set_xlabel('Mức độ ảnh hưởng (Lượng sai số MAE tăng lên khi xáo trộn biến)', fontsize=11)
ax.set_ylabel('Đặc trưng đầu vào', fontsize=11)

plt.tight_layout()
plt.savefig('hinh_3_11_feature_importance.png', dpi=300)
plt.show()
```
